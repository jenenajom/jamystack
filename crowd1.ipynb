```python
import os
from PIL import Image

for img_file in os.listdir("test_images"):
    img_path = f"test_images/{img_file}"
    img = Image.open(img_path)

    # Run YOLOv5 detection
    results = model(img)

    # Print the results
    print(f"📸 Image: {img_file}")
    print(results.pandas().xyxy[0][results.pandas().xyxy[0]['name'] == 'person'])  # only show 'person' detections

    # Show the image with bounding boxes
    results.show()
```

```python

import cv2
import numpy as np
import matplotlib.pyplot as plt

# --- Step 1: Select your video file ---
from google.colab import files
uploaded = files.upload()
video_path = list(uploaded.keys())[0]
print(f"Uploaded video: {video_path}")

# --- Step 2: Initialize video and models ---
cap = cv2.VideoCapture(video_path)
fgbg = cv2.createBackgroundSubtractorMOG2()
crowd_threshold = 100  # Set your overcrowding threshold
harassment_frames = []

frame_num = 0
while True:
    ret, frame = cap.read()
    if not ret:
        break
    frame_num += 1

    # --- Step 3: Crowd density estimation (replace with your model if available) ---
    # Dummy crowd count: use frame brightness as a proxy (replace with actual model!)
    crowd_count = int(np.sum(frame) / 1e6)

    # --- Step 4: Harassment detection (motion anomaly) ---
    fgmask = fgbg.apply(frame)
    movement = np.sum(fgmask > 200)
    harassment_detected = movement > 10000  # Tune threshold for sensitivity

    # --- Step 5: Alert and logging ---
    if crowd_count > crowd_threshold and harassment_detected:
        harassment_frames.append(frame_num)
        print(f"🚨 Frame {frame_num}: Overcrowding and possible harassment detected!")

cap.release()

# --- Step 6: Display results ---
print("Frames flagged for possible harassment during overcrowding:", harassment_frames)
cap = cv2.VideoCapture(video_path)
for fc in harassment_frames[:3]:  # Show first 3 flagged frames
    cap.set(cv2.CAP_PROP_POS_FRAMES, fc)
    ret, frame = cap.read()
    if ret:
        plt.imshow(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))
        plt.title(f"Flagged Frame {fc}")
        plt.axis('off')
        plt.show()
cap.release()
```
